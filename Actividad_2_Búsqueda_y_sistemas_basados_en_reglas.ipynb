El siguiente codigo muestra un algoritmo algoritmo A*  para encontrar la distacia mas corta entre dos puntos.

Como base de conocimiento se utilizo un mapa real del barrio tunal en la ciudad de Bogotá, en el cual resido. Y se muestra a continuación:

El codigo almacena la información de los diferentes puntos y distancias entre estos, y utiliza el algoritmo A* para hallar el camino mas corto entre dos puntos dados.

```python
import heapq

# ================================
# Base de conocimientos escrito en reglas lógicas
# ================================

nombre_estaciones = {
    "1": "Entrada Parque - Avenida Boyacá",
    "2": "Entrada Parque - Carrera 24",
    "3": "Centro Comercial Tunal",
    "4": "Centro infantil del parque",
    "5": "Entrada Parque - Carrera 19c",
    "6": "Hospital Tunal",
    "7": "Biblioteca Publica Tunal"
}

grafo = {
    "1": {"2": 5, "5": 5},
    "2": {"3": 5, "4": 2, "1": 5},
    "3": {"2": 5, "7": 10},
    "4": {"2": 2, "5": 7, "7": 5},
    "5": {"1": 5, "4": 7, "7":10},
    "6": {"7": 1},
    "7": {"3": 10, "4": 5, "5": 10, "6": 1}
}

# ================================
# Algoritmo A* (A estrella)
# ================================
def a_estrella(grafo, inicio, objetivo):
    cola = [(0, inicio)]
    visitados = set()
    padres = {inicio: None}
    costos = {inicio: 0}

    while cola:
        (costo, actual) = heapq.heappop(cola)

        if actual == objetivo:
            camino = []
            while actual is not None:
                camino.append(actual)
                actual = padres[actual]
            return camino[::-1], costos[objetivo]

        if actual in visitados:
            continue

        visitados.add(actual)

        for vecino, peso in grafo[actual].items():
            nuevo_costo = costos[actual] + peso
            if vecino not in costos or nuevo_costo < costos[vecino]:
                costos[vecino] = nuevo_costo
                prioridad = nuevo_costo
                heapq.heappush(cola, (prioridad, vecino))
                padres[vecino] = actual

    return None, float("inf")


def imprimir_camino(camino, costo_total):
  print("========== RESULTADO ==========")
  print(f"Ruta más corta desde {nombre_estaciones[inicio]} hasta {nombre_estaciones[objetivo]}:")
  if camino:
      tiempo_anterior = 0
      for i in range(len(camino)):
          estacion_actual_id = camino[i]
          estacion_actual_nombre = nombre_estaciones[estacion_actual_id]

          if i == 0:
              tiempo_desde_anterior = 0
          else:
              estacion_anterior_id = camino[i-1]
              tiempo_desde_anterior = grafo[estacion_anterior_id][estacion_actual_id]

          print(f"{estacion_actual_id}. {estacion_actual_nombre} (+{tiempo_desde_anterior} min)")

      print(f"\nTiempo total: {costo_total} minutos")
  else:
      print("No se encontró una ruta.")
```
